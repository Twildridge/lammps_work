```python
#!/usr/bin/env python3
"""
Add lateral walls to an equilibrated gel slab configuration.
- Reads equilibrated LAMMPS data file
- Rotates system (x-y plane only) to align gel with box axes using PCA
- Finds gel extent from polymer beads
- Adds wall atoms (type 6) on ±x and ±y faces
- Removes solvent outside the walls (keeps solvent above/below gel)
- Writes new data file with walls

Usage:
    python add_walls_to_slab.py <input_data_file> <output_data_file> [wall_spacing]
    
Example:
    python add_walls_to_slab.py final_config_slab.data slab_with_walls.data 0.4
"""

import numpy as np
import sys
import os


def rotate_to_align_gel(atoms):
    """
    Rotate all atoms in x–y plane so gel slab is axis-aligned.
    Uses PCA on polymer beads and chooses rotation that minimizes
    the x–y bounding box area of the gel.
    """

    polymer_atoms = [a for a in atoms if a['type'] in [1, 2]]
    if not polymer_atoms:
        raise ValueError("No polymer atoms found!")

    # Polymer coordinates
    xy = np.array([[a['x'], a['y']] for a in polymer_atoms])

    # Polymer COM (rotation center!)
    com = xy.mean(axis=0)
    centered = xy - com

    # PCA
    cov = np.cov(centered.T)
    eigvals, eigvecs = np.linalg.eigh(cov)
    principal = eigvecs[:, np.argmax(eigvals)]

    # Candidate angles (± ambiguity)
    angles = [
        np.arctan2(principal[1], principal[0]),
        np.arctan2(-principal[1], -principal[0])
    ]

    def bbox_area(theta):
        c, s = np.cos(-theta), np.sin(-theta)
        R = np.array([[c, -s], [s, c]])
        rot = centered @ R.T
        dx = rot[:, 0].max() - rot[:, 0].min()
        dy = rot[:, 1].max() - rot[:, 1].min()
        return dx * dy

    # Choose angle giving tightest slab
    angle = min(angles, key=bbox_area)

    print(f"  Gel rotation angle: {np.degrees(angle):.2f} degrees")

    c, s = np.cos(-angle), np.sin(-angle)

    # Rotate ALL atoms about polymer COM
    for a in atoms:
        x = a['x'] - com[0]
        y = a['y'] - com[1]

        a['x'] = c * x - s * y + com[0]
        a['y'] = s * x + c * y + com[1]
        # z unchanged

    return atoms, angle



def parse_lammps_data(filename):
    """Parse LAMMPS data file and extract all information."""
    
    atoms = []
    bonds = []
    box_bounds = {}
    masses = {}
    header_info = {}
    
    with open(filename, 'r') as f:
        lines = f.readlines()
    
    i = 0
    while i < len(lines):
        line = lines[i].strip()
        
        # Parse header info
        if 'atoms' in line and 'atom' not in line.lower().split()[0]:
            header_info['natoms'] = int(line.split()[0])
        elif 'bonds' in line and 'bond' not in line.lower().split()[0]:
            header_info['nbonds'] = int(line.split()[0])
        elif 'atom types' in line:
            header_info['atom_types'] = int(line.split()[0])
        elif 'bond types' in line:
            header_info['bond_types'] = int(line.split()[0])
        elif 'xlo xhi' in line:
            parts = line.split()
            box_bounds['xlo'] = float(parts[0])
            box_bounds['xhi'] = float(parts[1])
        elif 'ylo yhi' in line:
            parts = line.split()
            box_bounds['ylo'] = float(parts[0])
            box_bounds['yhi'] = float(parts[1])
        elif 'zlo zhi' in line:
            parts = line.split()
            box_bounds['zlo'] = float(parts[0])
            box_bounds['zhi'] = float(parts[1])
        
        # Parse Masses section
        elif line == 'Masses':
            i += 2  # Skip blank line
            while i < len(lines) and lines[i].strip() and not lines[i].strip().startswith(('Atoms', 'Bonds', 'Pair')):
                parts = lines[i].split()
                if len(parts) >= 2:
                    try:
                        masses[int(parts[0])] = float(parts[1])
                    except ValueError:
                        pass
                i += 1
            continue
        
        # Parse Atoms section
        elif line == 'Atoms' or line.startswith('Atoms'):
            i += 2  # Skip blank line
            while i < len(lines) and lines[i].strip() and not lines[i].strip().startswith(('Bonds', 'Velocities', 'Pair')):
                parts = lines[i].split()
                if len(parts) >= 6:
                    try:
                        atoms.append({
                            'id': int(parts[0]),
                            'mol': int(parts[1]),
                            'type': int(parts[2]),
                            'x': float(parts[3]),
                            'y': float(parts[4]),
                            'z': float(parts[5])
                        })
                    except ValueError:
                        pass
                i += 1
            continue
        
        # Parse Bonds section
        elif line == 'Bonds' or line.startswith('Bonds'):
            i += 2  # Skip blank line
            while i < len(lines) and lines[i].strip():
                parts = lines[i].split()
                if len(parts) >= 4:
                    try:
                        bonds.append({
                            'id': int(parts[0]),
                            'type': int(parts[1]),
                            'atom1': int(parts[2]),
                            'atom2': int(parts[3])
                        })
                    except ValueError:
                        pass
                i += 1
            continue
        
        i += 1
    
    return atoms, bonds, box_bounds, masses, header_info


def find_gel_extent(atoms):
    """Find the spatial extent of the gel from polymer beads (types 1, 2)."""
    
    polymer_atoms = [a for a in atoms if a['type'] in [1, 2]]
    
    if not polymer_atoms:
        raise ValueError("No polymer atoms (types 1, 2) found!")
    
    x_coords = [a['x'] for a in polymer_atoms]
    y_coords = [a['y'] for a in polymer_atoms]
    z_coords = [a['z'] for a in polymer_atoms]
    
    extent = {
        'xmin': min(x_coords),
        'xmax': max(x_coords),
        'ymin': min(y_coords),
        'ymax': max(y_coords),
        'zmin': min(z_coords),
        'zmax': max(z_coords)
    }
    
    return extent


def generate_wall_atoms(gel_extent, box_bounds, spacing=0.4):
    """Generate wall atoms on ±x and ±y faces using hexagonal packing."""
    
    wall_atoms = []
    
    # Wall positions (at gel boundary)
    x_min_wall = gel_extent['xmin']
    x_max_wall = gel_extent['xmax']
    y_min_wall = gel_extent['ymin']
    y_max_wall = gel_extent['ymax']
    
    # Z range: full box height
    z_lo = box_bounds['zlo']
    z_hi = box_bounds['zhi']
    
    # Hexagonal packing parameters
    dz = spacing
    dy_hex = spacing * np.sqrt(3)
    dx_hex = spacing * np.sqrt(3)
    
    # -X wall (slab in y-z plane at x = x_min_wall)
    nz = int((z_hi - z_lo) / dz) + 1
    ny = int((y_max_wall - y_min_wall) / dy_hex) + 1
    
    for j in range(ny):
        for k in range(nz):
            y = y_min_wall + j * dy_hex + (spacing / 2 if k % 2 else 0)
            z = z_lo + k * dz
            
            if y_min_wall <= y <= y_max_wall:
                wall_atoms.append({
                    'type': 6,
                    'x': x_min_wall,
                    'y': y,
                    'z': z
                })
    
    # +X wall (slab in y-z plane at x = x_max_wall)
    for j in range(ny):
        for k in range(nz):
            y = y_min_wall + j * dy_hex + (spacing / 2 if k % 2 else 0)
            z = z_lo + k * dz
            
            if y_min_wall <= y <= y_max_wall:
                wall_atoms.append({
                    'type': 6,
                    'x': x_max_wall,
                    'y': y,
                    'z': z
                })
    
    # -Y wall (slab in x-z plane at y = y_min_wall)
    nx = int((x_max_wall - x_min_wall) / dx_hex) + 1
    
    for i in range(nx):
        for k in range(nz):
            x = x_min_wall + i * dx_hex + (spacing / 2 if k % 2 else 0)
            z = z_lo + k * dz
            
            if x_min_wall <= x <= x_max_wall:
                wall_atoms.append({
                    'type': 6,
                    'x': x,
                    'y': y_min_wall,
                    'z': z
                })
    
    # +Y wall (slab in x-z plane at y = y_max_wall)
    for i in range(nx):
        for k in range(nz):
            x = x_min_wall + i * dx_hex + (spacing / 2 if k % 2 else 0)
            z = z_lo + k * dz
            
            if x_min_wall <= x <= x_max_wall:
                wall_atoms.append({
                    'type': 6,
                    'x': x,
                    'y': y_max_wall,
                    'z': z
                })
    
    return wall_atoms


def remove_external_solvent(atoms, gel_extent):
    """Remove solvent atoms outside the wall boundaries (but keep above/below gel)."""
    
    x_min = gel_extent['xmin']
    x_max = gel_extent['xmax']
    y_min = gel_extent['ymin']
    y_max = gel_extent['ymax']
    
    filtered_atoms = []
    removed_count = 0
    
    for atom in atoms:
        if atom['type'] == 3:  # Solvent
            # Remove if outside x or y bounds
            if atom['x'] < x_min or atom['x'] > x_max or \
               atom['y'] < y_min or atom['y'] > y_max:
                removed_count += 1
                continue
        
        filtered_atoms.append(atom)
    
    print(f"Removed {removed_count} solvent atoms outside walls")
    return filtered_atoms


def write_lammps_data(filename, atoms, bonds, box_bounds, masses):
    """Write LAMMPS data file with updated atoms and bonds."""
    
    # Create mapping from old atom IDs to new IDs
    old_to_new_id = {}
    for new_id, atom in enumerate(atoms, start=1):
        old_to_new_id[atom['id']] = new_id
    
    # Update bond references and filter out broken bonds
    valid_bonds = []
    for bond in bonds:
        if bond['atom1'] in old_to_new_id and bond['atom2'] in old_to_new_id:
            valid_bonds.append({
                'id': len(valid_bonds) + 1,
                'type': bond['type'],
                'atom1': old_to_new_id[bond['atom1']],
                'atom2': old_to_new_id[bond['atom2']]
            })
    
    # Get max molecule ID for wall atoms
    max_mol = max(a['mol'] for a in atoms if 'mol' in a)
    
    with open(filename, 'w') as f:
        f.write("LAMMPS data file with lateral walls\n\n")
        f.write(f"{len(atoms)} atoms\n")
        f.write(f"{len(valid_bonds)} bonds\n")
        f.write("0 angles\n")
        f.write("0 dihedrals\n")
        f.write("0 impropers\n\n")
        f.write("6 atom types\n")
        f.write("1 bond types\n\n")
        f.write(f"{box_bounds['xlo']} {box_bounds['xhi']} xlo xhi\n")
        f.write(f"{box_bounds['ylo']} {box_bounds['yhi']} ylo yhi\n")
        f.write(f"{box_bounds['zlo']} {box_bounds['zhi']} zlo zhi\n\n")
        
        f.write("Masses\n\n")
        for i in range(1, 7):
            mass = masses.get(i, 1.0)
            f.write(f"{i} {mass}\n")
        f.write("\n")
        
        f.write("Atoms\n\n")
        for new_id, atom in enumerate(atoms, start=1):
            mol = atom.get('mol', max_mol + new_id)
            f.write(f"{new_id} {mol} {atom['type']} {atom['x']:.6f} {atom['y']:.6f} {atom['z']:.6f}\n")
        
        if valid_bonds:
            f.write("\nBonds\n\n")
            for bond in valid_bonds:
                f.write(f"{bond['id']} {bond['type']} {bond['atom1']} {bond['atom2']}\n")
    
    return len(valid_bonds)


def add_walls_to_slab(input_file, output_file, wall_spacing=0.4):
    """Main function to add walls to equilibrated slab."""
    
    print(f"Reading {input_file}...")
    atoms, bonds, box_bounds, masses, header_info = parse_lammps_data(input_file)
    
    print(f"  Original atoms: {len(atoms)}")
    print(f"  Original bonds: {len(bonds)}")
    
    # Count atom types
    type_counts = {}
    for a in atoms:
        t = a['type']
        type_counts[t] = type_counts.get(t, 0) + 1
    print(f"  Atom types: {type_counts}")
    
    # Rotate to align gel with box axes
    print(f"\nRotating system to align gel with box axes...")
    atoms, rotation_angle = rotate_to_align_gel(atoms)
    
    # Find gel extent (now axis-aligned)
    gel_extent = find_gel_extent(atoms)
    print(f"\nGel extent (after rotation):")
    print(f"  X: {gel_extent['xmin']:.2f} to {gel_extent['xmax']:.2f} (width: {gel_extent['xmax']-gel_extent['xmin']:.2f})")
    print(f"  Y: {gel_extent['ymin']:.2f} to {gel_extent['ymax']:.2f} (width: {gel_extent['ymax']-gel_extent['ymin']:.2f})")
    print(f"  Z: {gel_extent['zmin']:.2f} to {gel_extent['zmax']:.2f} (height: {gel_extent['zmax']-gel_extent['zmin']:.2f})")
    
    # Remove external solvent
    print(f"\nRemoving solvent outside gel bounds...")
    atoms = remove_external_solvent(atoms, gel_extent)
    
    # Generate wall atoms
    print(f"\nGenerating wall atoms (spacing={wall_spacing})...")
    wall_atoms = generate_wall_atoms(gel_extent, box_bounds, spacing=wall_spacing)
    print(f"  Generated {len(wall_atoms)} wall atoms")
    
    # Get max molecule ID and atom ID
    max_mol = max(a['mol'] for a in atoms)
    max_id = max(a['id'] for a in atoms)
    
    # Add wall atoms to atom list
    for i, wall in enumerate(wall_atoms):
        atoms.append({
            'id': max_id + i + 1,
            'mol': max_mol + i + 1,
            'type': wall['type'],
            'x': wall['x'],
            'y': wall['y'],
            'z': wall['z']
        })
    
    # Write output
    print(f"\nWriting {output_file}...")
    final_bonds = write_lammps_data(output_file, atoms, bonds, box_bounds, masses)
    
    # Final summary
    type_counts_final = {}
    for a in atoms:
        t = a['type']
        type_counts_final[t] = type_counts_final.get(t, 0) + 1
    
    print(f"\nFinal configuration:")
    print(f"  Total atoms: {len(atoms)}")
    print(f"  Total bonds: {final_bonds}")
    print(f"  Atom types: {type_counts_final}")
    print(f"  Wall atoms (type 6): {type_counts_final.get(6, 0)}")
    print(f"\nOutput written to {output_file}")

```

```python
# Inputs
input_file = "../lammps_data/final_config_slab_support_5beads_10x10x5_rho6_extra_padding43_1.5_1.4_10000000.data"
output_file = "../lammps_data/walled_slab_support_5beads_10x10x5_rho6_extra_padding43_1.5_1.4_10000000.data"
wall_spacing = 0.1

add_walls_to_slab(input_file, output_file, wall_spacing)
```

```python
#!/usr/bin/env python3
"""
Add four vertical (z-normal) lateral walls to an equilibrated gel slab.

This script is fully consistent with the following LAMMPS group definitions:

    group polymer type 1 2
    group solvent type 3
    group support type 4
    group piston  type 5
    group walls   type 6
    group mobile  type 1 2 3
    group gel_system type 1 2 4 5 6

What this script does:
- Rotates ONLY the mobile group (types 1,2,3) in the x–y plane so the gel slab
  edges are square with the piston/support (types 4,5 are not rotated)
- Determines gel lateral extent from polymer beads with a clearance margin
- Removes solvent outside the lateral walls (keeps solvent above/below the gel)
- Generates four vertical walls (type 6) normal to z
- Writes a new LAMMPS data file
"""

import numpy as np


# =========================
# Group-consistent settings
# =========================
POLYMER_TYPES = {1, 2}        # group polymer
SOLVENT_TYPES = {3}           # group solvent
SUPPORT_TYPES = {4}           # group support
PISTON_TYPES  = {5}           # group piston
WALL_TYPE     = 6             # group walls

ROTATE_TYPES  = POLYMER_TYPES | SOLVENT_TYPES   # group mobile
WALL_CLEARANCE = 0.2                            # gap between gel and walls


# =========================
# Geometry utilities
# =========================
def find_min_bounding_rect_angle(hull_points):
    """
    Find the rotation angle that gives the minimum area bounding rectangle.
    Uses rotating calipers on convex hull.
    """
    n = len(hull_points)
    min_area = float('inf')
    best_angle = 0
    
    for i in range(n):
        # Edge vector
        edge = hull_points[(i + 1) % n] - hull_points[i]
        edge_angle = np.arctan2(edge[1], edge[0])
        
        # Rotate hull points to align this edge with x-axis
        c, s = np.cos(-edge_angle), np.sin(-edge_angle)
        rotated = np.column_stack([
            c * hull_points[:, 0] - s * hull_points[:, 1],
            s * hull_points[:, 0] + c * hull_points[:, 1]
        ])
        
        # Get bounding box area in rotated frame
        width = rotated[:, 0].max() - rotated[:, 0].min()
        height = rotated[:, 1].max() - rotated[:, 1].min()
        area = width * height
        
        if area < min_area:
            min_area = area
            best_angle = edge_angle
    
    return best_angle


def rotate_polymer_and_solvent(atoms):
    """
    Rotate mobile group (types 1,2,3) to align minimum bounding rectangle with axes.
    """
    poly = [a for a in atoms if a['type'] in POLYMER_TYPES]
    if not poly:
        raise ValueError("No polymer atoms found")

    xy = np.array([[a['x'], a['y']] for a in poly])
    com = xy.mean(axis=0)

    # Find minimum bounding rectangle angle
    hull = ConvexHull(xy)
    hull_points = xy[hull.vertices]
    angle = find_min_bounding_rect_angle(hull_points)

    print(f"Slab rotation angle (MABR): {np.degrees(angle):.2f} degrees")

    c, s = np.cos(-angle), np.sin(-angle)

    for a in atoms:
        if a['type'] in ROTATE_TYPES:
            x = a['x'] - com[0]
            y = a['y'] - com[1]
            a['x'] = c*x - s*y + com[0]
            a['y'] = s*x + c*y + com[1]

    return atoms


from scipy.spatial import ConvexHull

def remove_rogue_polymer(atoms, percentile=0.5):
    """Remove polymer beads that are outliers in x-y plane."""
    poly = [a for a in atoms if a['type'] in POLYMER_TYPES]
    xs = np.array([a['x'] for a in poly])
    ys = np.array([a['y'] for a in poly])
    
    xmin, xmax = np.percentile(xs, percentile), np.percentile(xs, 100 - percentile)
    ymin, ymax = np.percentile(ys, percentile), np.percentile(ys, 100 - percentile)
    
    kept = []
    removed = 0
    for a in atoms:
        if a['type'] in POLYMER_TYPES:
            if a['x'] < xmin or a['x'] > xmax or a['y'] < ymin or a['y'] > ymax:
                removed += 1
                continue
        kept.append(a)
    
    print(f"Removed {removed} rogue polymer atoms")
    return kept

def find_gel_extent(atoms, clearance, percentile=0.05):
    """Gel extent from polymer beads, using percentiles to exclude rogues."""
    poly = [a for a in atoms if a['type'] in POLYMER_TYPES]
    
    xs = np.array([a['x'] for a in poly])
    ys = np.array([a['y'] for a in poly])
    zs = np.array([a['z'] for a in poly])

    # Use percentiles to exclude outliers
    xmin = np.percentile(xs, percentile)
    xmax = np.percentile(xs, 100 - percentile)
    ymin = np.percentile(ys, percentile)
    ymax = np.percentile(ys, 100 - percentile)

    print(f"Polymer x range: {xmin:.2f} to {xmax:.2f} (width: {xmax-xmin:.2f})")
    print(f"Polymer y range: {ymin:.2f} to {ymax:.2f} (width: {ymax-ymin:.2f})")

    return {
        'xmin': xmin - clearance,
        'xmax': xmax + clearance,
        'ymin': ymin - clearance,
        'ymax': ymax + clearance,
        'zmin': zs.min(),
        'zmax': zs.max()
    }


def generate_wall_atoms(ext, box, spacing=0.2):
    """Generate four vertical walls (type 6) as single-layer triangular lattices."""
    walls = []
    
    zlo, zhi = box['zlo'], box['zhi']
    
    # For triangular lattice: rows are offset by spacing/2, row spacing is spacing*sqrt(3)/2
    row_spacing = spacing * np.sqrt(3) / 2
    
    # ±x walls (constant x, triangular lattice in y-z plane)
    for x in [ext['xmin'], ext['xmax']]:
        ny = int(np.ceil((ext['ymax'] - ext['ymin']) / spacing)) + 1
        nz = int(np.ceil((zhi - zlo) / row_spacing)) + 1
        
        for iz in range(nz):
            z = zlo + iz * row_spacing
            if z > zhi:
                continue
                
            # Alternate rows offset by spacing/2
            y_offset = (spacing / 2) if (iz % 2) else 0
            y_start = ext['ymin'] + y_offset
            
            y = y_start
            while y <= ext['ymax']:
                walls.append({'type': WALL_TYPE, 'x': x, 'y': y, 'z': z})
                y += spacing
    
    # ±y walls (constant y, triangular lattice in x-z plane)
    for y in [ext['ymin'], ext['ymax']]:
        nx = int(np.ceil((ext['xmax'] - ext['xmin']) / spacing)) + 1
        nz = int(np.ceil((zhi - zlo) / row_spacing)) + 1
        
        for iz in range(nz):
            z = zlo + iz * row_spacing
            if z > zhi:
                continue
                
            # Alternate rows offset by spacing/2
            x_offset = (spacing / 2) if (iz % 2) else 0
            x_start = ext['xmin'] + x_offset
            
            x = x_start
            while x <= ext['xmax']:
                walls.append({'type': WALL_TYPE, 'x': x, 'y': y, 'z': z})
                x += spacing
    
    print(f"Generated {len(walls)} wall atoms with triangular packing (spacing={spacing})")
    return walls



def remove_external_solvent(atoms, ext):
    """Remove solvent outside lateral wall bounds."""
    kept = []
    removed = 0

    for a in atoms:
        if a['type'] in SOLVENT_TYPES:
            if (a['x'] < ext['xmin'] or a['x'] > ext['xmax'] or
                a['y'] < ext['ymin'] or a['y'] > ext['ymax']):
                removed += 1
                continue
        kept.append(a)

    print(f"Removed {removed} solvent atoms")
    return kept

def remove_external_polymer(atoms, ext):
    """Remove polymer atoms outside wall boundaries."""
    kept = []
    removed = 0
    
    for a in atoms:
        if a['type'] in POLYMER_TYPES:
            if (a['x'] < ext['xmin'] or a['x'] > ext['xmax'] or
                a['y'] < ext['ymin'] or a['y'] > ext['ymax']):
                removed += 1
                continue
        kept.append(a)
    
    print(f"Removed {removed} polymer atoms outside walls")
    return kept

# =========================
# LAMMPS I/O
# =========================
def parse_lammps_data(filename):
    """Parse LAMMPS data file and extract all information."""
    
    atoms = []
    bonds = []
    box_bounds = {}
    masses = {}
    header_info = {}
    
    with open(filename, 'r') as f:
        lines = f.readlines()
    
    i = 0
    while i < len(lines):
        line = lines[i].strip()
        
        # Parse header info
        if 'atoms' in line and 'atom' not in line.lower().split()[0]:
            header_info['natoms'] = int(line.split()[0])
        elif 'bonds' in line and 'bond' not in line.lower().split()[0]:
            header_info['nbonds'] = int(line.split()[0])
        elif 'atom types' in line:
            header_info['atom_types'] = int(line.split()[0])
        elif 'bond types' in line:
            header_info['bond_types'] = int(line.split()[0])
        elif 'xlo xhi' in line:
            parts = line.split()
            box_bounds['xlo'] = float(parts[0])
            box_bounds['xhi'] = float(parts[1])
        elif 'ylo yhi' in line:
            parts = line.split()
            box_bounds['ylo'] = float(parts[0])
            box_bounds['yhi'] = float(parts[1])
        elif 'zlo zhi' in line:
            parts = line.split()
            box_bounds['zlo'] = float(parts[0])
            box_bounds['zhi'] = float(parts[1])
        
        # Parse Masses section
        elif line == 'Masses':
            i += 2  # Skip blank line
            while i < len(lines) and lines[i].strip() and not lines[i].strip().startswith(('Atoms', 'Bonds', 'Pair')):
                parts = lines[i].split()
                if len(parts) >= 2:
                    try:
                        masses[int(parts[0])] = float(parts[1])
                    except ValueError:
                        pass
                i += 1
            continue
        
        # Parse Atoms section
        elif line == 'Atoms' or line.startswith('Atoms'):
            i += 2  # Skip blank line
            while i < len(lines) and lines[i].strip() and not lines[i].strip().startswith(('Bonds', 'Velocities', 'Pair')):
                parts = lines[i].split()
                if len(parts) >= 6:
                    try:
                        atoms.append({
                            'id': int(parts[0]),
                            'mol': int(parts[1]),
                            'type': int(parts[2]),
                            'x': float(parts[3]),
                            'y': float(parts[4]),
                            'z': float(parts[5])
                        })
                    except ValueError:
                        pass
                i += 1
            continue
        
        # Parse Bonds section
        elif line == 'Bonds' or line.startswith('Bonds'):
            i += 2  # Skip blank line
            while i < len(lines) and lines[i].strip():
                parts = lines[i].split()
                if len(parts) >= 4:
                    try:
                        bonds.append({
                            'id': int(parts[0]),
                            'type': int(parts[1]),
                            'atom1': int(parts[2]),
                            'atom2': int(parts[3])
                        })
                    except ValueError:
                        pass
                i += 1
            continue
        
        i += 1
    
    return atoms, bonds, box_bounds, masses


def write_lammps_data(filename, atoms, bonds, box, masses):
    old2new = {a['id']: i+1 for i,a in enumerate(atoms)}

    valid_bonds = [
        {'id': i+1, 'type': b['type'],
         'atom1': old2new[b['atom1']], 'atom2': old2new[b['atom2']]}
        for i,b in enumerate(bonds)
        if b['atom1'] in old2new and b['atom2'] in old2new
    ]

    with open(filename, 'w') as f:
        f.write("LAMMPS data file with lateral walls\n\n")
        f.write(f"{len(atoms)} atoms\n{len(valid_bonds)} bonds\n\n")
        f.write("6 atom types\n1 bond types\n\n")
        f.write(f"{box['xlo']} {box['xhi']} xlo xhi\n")
        f.write(f"{box['ylo']} {box['yhi']} ylo yhi\n")
        f.write(f"{box['zlo']} {box['zhi']} zlo zhi\n\n")

        f.write("Masses\n\n")
        for i in range(1,7):
            f.write(f"{i} {masses.get(i,1.0)}\n")

        f.write("\nAtoms\n\n")
        for i,a in enumerate(atoms,1):
            f.write(f"{i} {a['mol']} {a['type']} "
                    f"{a['x']:.6f} {a['y']:.6f} {a['z']:.6f}\n")

        if valid_bonds:
            f.write("\nBonds\n\n")
            for b in valid_bonds:
                f.write(f"{b['id']} {b['type']} {b['atom1']} {b['atom2']}\n")


# =========================
# Main driver
# =========================
def add_walls_to_slab(input_file, output_file, wall_spacing):
    atoms, bonds, box, masses = parse_lammps_data(input_file)

    atoms = rotate_polymer_and_solvent(atoms)
    ext = find_gel_extent(atoms, WALL_CLEARANCE, percentile=0.1)
    
    # Remove atoms outside walls
    atoms = remove_external_polymer(atoms, ext)
    atoms = remove_external_solvent(atoms, ext)

    walls = generate_wall_atoms(ext, box, wall_spacing)

    max_id = max(a['id'] for a in atoms)
    max_mol = max(a['mol'] for a in atoms)

    for i, w in enumerate(walls):
        atoms.append({
            'id': max_id + i + 1,
            'mol': max_mol + i + 1,
            'type': WALL_TYPE,
            'x': w['x'], 'y': w['y'], 'z': w['z']
        })

    write_lammps_data(output_file, atoms, bonds, box, masses)
    print(f"Wrote {output_file}")
```

```python
# Inputs

input_file = "../lammps_data/final_config_slab_support_5beads_10x10x5_rho6_extra_padding43_1.5_1.4_10000000.data"
output_file = "../lammps_data/walled_slab_support_5beads_10x10x5_rho6_extra_padding43_1.5_1.4_10000000.data"
wall_spacing = 0.2

add_walls_to_slab(input_file, output_file, wall_spacing)

```
